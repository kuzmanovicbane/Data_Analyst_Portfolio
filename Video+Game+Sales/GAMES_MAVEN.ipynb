```python
import pandas as pd
```

```python
games = pd.read_csv('vgchartz-2024.csv')
```

```python
games
```

```python
games.columns
```

```python
games.info()
```

```python
games['release_date'] = pd.to_datetime(games['release_date'], errors='coerce', dayfirst=False)
```

```python
games['relese_year'] = games['release_date'].dt.year
games['relese_month'] = games['release_date'].dt.month
```

```python
games['release_date'].dtype
```

```python
games.head()
```

```python
games['relese_year'] = games['relese_year'].astype('Int64')
games['relese_month'] = games['relese_month'].astype('Int64')
games.head()
```

```python
title_group = games.groupby(['title','relese_year','relese_month']).agg({'critic_score':'mean', 'total_sales':'sum', 'na_sales':'sum', 'jp_sales':'sum', 'pal_sales':'sum', 'other_sales':'sum'}).sort_values(['total_sales'], ascending = False)
title_group
```

```python
year_month_group = games.groupby(['relese_year','relese_month']).agg({'critic_score':'mean', 'total_sales':'sum', 'na_sales':'sum', 'jp_sales':'sum', 'pal_sales':'sum', 'other_sales':'sum'}).sort_values(['total_sales'], ascending = False)
year_month_group
```

```python
Pivot_Console_By_Genre = games.pivot_table(index='console',columns='genre', values='title', aggfunc = 'count')
Pivot_Console_By_Genre
```

```python
with pd.ExcelWriter('GAMES_from_Python.xlsx', engine='openpyxl') as writer:
    title_group.to_excel(writer, sheet_name='titles_per_year_months')
    Pivot_Console_By_Genre.to_excel(writer, sheet_name='ConsolesByGenres')
```
