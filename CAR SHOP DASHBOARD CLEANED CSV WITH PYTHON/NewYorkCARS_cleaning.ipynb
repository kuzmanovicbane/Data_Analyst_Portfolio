```python
import pandas as pd
import matplotlib as mt
```

```python
CarRates = pd.read_csv('Car_Rates.csv')
NewYorkCars = pd.read_csv('New_York_cars.csv')
```

```python
Average_Num_of_reviews = CarRates['Num_of_reviews'].mean()
Average_Num_of_reviews
```

```python
Average_General_rate = CarRates['General_rate'].mean()
Average_General_rate
```

```python
Average_Comfort = CarRates['Comfort'].mean()
Average_Comfort
```

```python
Average_Interior_design = CarRates['Interior design'].mean()
Average_Interior_design
```

```python
Average_Performance = CarRates['Performance'].mean()
Average_Performance
```

```python
Average_Value = CarRates['Value for the money'].mean()
Average_Value
```

```python
Average_Exterior_styling = CarRates['Exterior styling'].mean()
Average_Exterior_styling
```

```python
Average_Reliability = CarRates['Reliability'].mean()
Average_Reliability
```

```python
CarRates['Num_of_reviews'] = CarRates['Num_of_reviews'].fillna(Average_Num_of_reviews)
CarRates['General_rate'] = CarRates['General_rate'].fillna(Average_General_rate)
CarRates['Comfort'] = CarRates['Comfort'].fillna(Average_Comfort)
CarRates['Interior design'] = CarRates['Interior design'].fillna(Average_Interior_design)
CarRates['Performance'] = CarRates['Performance'].fillna(Average_Performance)
CarRates['Value for the money'] = CarRates['Value for the money'].fillna(Average_Value)
CarRates['Exterior styling'] = CarRates['Exterior styling'].fillna(Average_Exterior_styling)
CarRates['Reliability'] = CarRates['Reliability'].fillna(Average_Reliability)

CarRates.isna().sum()
```

```python
NewYorkCars['Exterior color'] = NewYorkCars['Exterior color'].fillna('NOT DEFINED')
NewYorkCars['Interior color'] = NewYorkCars['Interior color'].fillna('NOT DEFINED')
NewYorkCars['Drivetrain'] = NewYorkCars['Drivetrain'].fillna('NOT DEFINED')
NewYorkCars['MPG'] = NewYorkCars['MPG'].fillna(NewYorkCars['MPG'].mode()[0])
NewYorkCars['Fuel type'] = NewYorkCars['Fuel type'].fillna(NewYorkCars['Fuel type'].mode()[0])
NewYorkCars['Transmission'] = NewYorkCars['Transmission'].fillna(NewYorkCars['Transmission'].mode()[0])
NewYorkCars['Engine'] = NewYorkCars['Engine'].fillna('NOT DEFINED')
NewYorkCars['Mileage'] = NewYorkCars['Mileage'].fillna(NewYorkCars['Mileage'].mean())
NewYorkCars['Convenience'] = NewYorkCars['Convenience'].fillna('WITHOUT CONVENIENCE')
NewYorkCars['Entertainment'] = NewYorkCars['Entertainment'].fillna('WITHOUT ENTERTAINMENT')
NewYorkCars['Exterior'] = NewYorkCars['Exterior'].fillna('WITHOUT SPECIFIC EXTERIOR ADDINGS')
NewYorkCars['Safety'] =  NewYorkCars['Safety'].fillna('SPECIFICATION IS NOT DEFINED')
NewYorkCars['Seating'] = NewYorkCars['Seating'].fillna('Basic Seats')
NewYorkCars['Accidents or damage'] = NewYorkCars['Accidents or damage'].fillna('WITHOUT ACCIDENTS OR DAMAGE')
NewYorkCars['Clean title'] = NewYorkCars['Clean title'].fillna('No')
NewYorkCars['1-owner vehicle'] = NewYorkCars['1-owner vehicle'].fillna('unknown')
NewYorkCars['Personal use only'] = NewYorkCars['Personal use only'].fillna('No')

NewYorkCars.isna().sum()
```
