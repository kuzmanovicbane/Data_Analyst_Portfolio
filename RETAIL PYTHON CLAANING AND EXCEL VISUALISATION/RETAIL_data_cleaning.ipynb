```python
# Funkcija koja popunjava NaN u ShortItemName najčešćom vrednošću unutar svake kategorije
def fill_short_item_name_with_mode(group):
    mode_value = group['ShortItemName'].mode()
    if not mode_value.empty:
        group['ShortItemName'] = group['ShortItemName'].fillna(mode_value[0])
    return group

# Primeni funkciju po grupama 'Category'
retailSTORES = retailSTORES.groupby('Category').apply(fill_short_item_name_with_mode)
retailSTORES
```

```python
# Pronađi najčešću vrednost ShortItemName po Category
mode_per_category = retailSTORES.groupby('Category')['ShortItemName'].agg(lambda x: x.mode().iloc[0])

# Mapiraj najčešću vrednost nazad na originalni DataFrame i postavi kolonu ShortItemName
retailSTORES['ShortItemName'] = retailSTORES['Category'].map(mode_per_category)

retailSTORES
```

```python
def fill_empty_price(row):
    if pd.isna(row['Price Per Unit']):
        if row['Quantity'] != 0 and not pd.isna(row['Quantity']):
            return row['Total Spent'] / row['Quantity']
        else:
            return 0  # ili neka druga vrednost po želji ako Quantity nije validan
    else:
        return row['Price Per Unit']

retailSTORES['Price Per Unit'] = retailSTORES.apply(fill_empty_price, axis=1)
```

```python
retailSTORES = retailSTORES.dropna(subset=['Quantity'])
retailSTORES.isna().sum()
```

```python
retailSTORES['Transaction Date'] = pd.to_datetime(retailSTORES['Transaction Date'], errors='coerce')
retailSTORES['Month'] = retailSTORES['Transaction Date'].dt.month
retailSTORES['Year'] = retailSTORES['Transaction Date'].dt.year
retailSTORES
```

```python
Pivot_Category = retailSTORES.pivot_table(
    index=['Category', 'Year', 'Month'],    
    values=['Quantity', 'Total Spent', 'Price Per Unit','Customer ID'],
    aggfunc={
        'Quantity': 'sum',
        'Total Spent': 'sum',
        'Price Per Unit': 'mean',
        'Customer ID': 'count'
    }
)
Pivot_Category
```

```python
Pivot_Payment_Method = retailSTORES.pivot_table(
    index=['Payment Method', 'Year', 'Month'],    
    values=['Quantity', 'Total Spent', 'Price Per Unit', 'Customer ID'],
    aggfunc={
        'Quantity': 'sum',
        'Total Spent': 'sum',
        'Price Per Unit': 'mean',
        'Customer ID': 'count'
    }
)

Pivot_Payment_Method
```

```python
Pivot_Location = retailSTORES.pivot_table(
    index=['Location', 'Year', 'Month'],    
    values=['Quantity', 'Total Spent', 'Price Per Unit', 'Customer ID'],
    aggfunc={
        'Quantity': 'sum',
        'Total Spent': 'sum',
        'Price Per Unit': 'mean',
        'Customer ID': 'count'
    }
)

Pivot_Location
```

```python
with pd.ExcelWriter('RETAIL_python_pivots.xlsx', engine = 'openpyxl') as writer:
    Pivot_Category.to_excel(writer, sheet_name='Pivot_Category')
    Pivot_Payment_Method.to_excel(writer, sheet_name='Pivot_Payment_Method')
    Pivot_Location.to_excel(writer, sheet_name=' Pivot_Location')
```
