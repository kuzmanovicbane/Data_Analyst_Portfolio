```python
import pandas as pd
```

```python
zomato = pd.read_csv('zomato.csv')
zomato
```

```python
zomato.columns
```

```python
zomato.info()
```

```python
zomato['restaurant type'].unique()
```

```python
zomato['restaurant type'] = zomato['restaurant type'].str.split(', ')
zomato['restaurant type']
```

```python
zomato_exploded = zomato.explode('restaurant type')
zomato_exploded
```

```python
zomato_exploded['restaurant type'].unique()
```

```python
zomato_exploded['cuisines type'] = zomato_exploded['cuisines type'].str.split(', ')
zomato_exploded['cuisines type']
```

```python
zomato_exploded2 = zomato_exploded.explode('cuisines type')
zomato_exploded3 = zomato_exploded2.explode('restaurant type')
zomato_exploded3
```

```python
zomato_exploded2['cuisines type'].unique()
```

```python
median_rate = zomato_exploded3['rate (out of 5)'].median()
median_rate
```

```python
zomato_exploded3['rate (out of 5)'] = zomato_exploded3['rate (out of 5)'].fillna(median_rate)
zomato_exploded3.isna().sum()
```

```python
mean_of_avg_cost = zomato_exploded3['avg cost (two people)'].mean().round(1)
mean_of_avg_cost
```

```python
zomato_exploded3['avg cost (two people)'] = zomato_exploded3['avg cost (two people)'].fillna(mean_of_avg_cost)
zomato_exploded3.isna().sum()
```

```python
min_rate = zomato_exploded3['rate (out of 5)'].min()
max_rate = zomato_exploded3['rate (out of 5)'].max()
```

```python
min_rate
```

```python
max_rate
```

```python
def group_rate (rate):
    if rate < 3:
        return "LOW RATING"
    if rate < 3.6:
        return "AWERAGE RATING"
    if rate < 4.3:
        return "GOOD RATING"
    else: 
        return "EXCELENT RATING"
        
zomato_exploded3['RATING GROUPS'] = zomato_exploded3['rate (out of 5)'].apply(group_rate)
```

```python
zomato_exploded3
```

```python
min_cost = zomato_exploded3['avg cost (two people)'].min()
max_cost = zomato_exploded3['avg cost (two people)'].max()
```

```python
min_cost
```

```python
max_cost
```

```python
median_cost = zomato_exploded3['avg cost (two people)'].median()
median_cost
```

```python
def cost_group (cost):
    if cost < 250:
        return "VERY CHEAP"
    if cost < 500:
        return "CHEAP"
    if cost < 1000:
        return "MODERATE"
    if cost < 2500:
        return "EXPENSIVE"
    if cost < 4000:
        return "LUXURY"
    else:
        return "ULTRA LUXARY"

zomato_exploded3['AVERAGE_COST_GROUPS'] = zomato_exploded3['avg cost (two people)'].apply(cost_group)
zomato_exploded3
```

```python
zomato_exploded3.to_csv('zomoto_clean_groups.csv', index = False)
```
