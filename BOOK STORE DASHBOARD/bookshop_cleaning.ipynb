```python
import pandas as pd
```

```python
bookshop = pd.read_csv('blackwell_shop_raw.csv')
bookshop.head()
```

```python
bookshop.info()
```

```python
bookshop.columns = bookshop.columns.str.strip()
bookshop['publication_date'] = pd.to_datetime(bookshop['publication_date'], errors='coerce')
```

```python
bookshop['type'].isna().sum()
```

```python
bookshop['type']=bookshop['type'].fillna("UNKNOWN")
bookshop['type'].isna().sum()
```

```python
bookshop['subtitle']=bookshop['subtitle'].fillna("WITHOUT SUBTITLE")
bookshop['subtitle'].isna().sum()
```

```python
bookshop['edition']=bookshop['edition'].fillna("NOT DEFINED")
bookshop['edition'].isna().sum()
```

```python
bookshop['author']=bookshop['author'].fillna("UNKNOWN")
bookshop['author'].isna().sum()
```

```python
bookshop['discount']=bookshop['discount'].fillna(0)
bookshop['discount'].isna().sum()
```

```python
bookshop['publisher']=bookshop['publisher'].fillna("UNKNOWN")
bookshop['publisher'].isna().sum()
```

```python
bookshop['published_country']=bookshop['published_country'].fillna("UNKNOWN")
bookshop['published_country'].isna().sum()
```

```python
bookshop['imprint']=bookshop['imprint'].fillna("UNKNOWN")
bookshop['imprint'].isna().sum()
```

```python
bookshop['language']=bookshop['language'].fillna("UNKNOWN")
bookshop['language'].isna().sum()
```

```python
# Uklanjanje svega osim cifara (0–9) ### CHAT GPT
bookshop['no_of_pages'] = bookshop['no_of_pages'].astype(str).str.replace(r'\D+', '', regex=True)
bookshop['no_of_pages'] = pd.to_numeric(bookshop['no_of_pages'], errors='coerce')
```

```python
mean_no_of_pages = bookshop['no_of_pages'].mean()
bookshop['no_of_pages']=bookshop['no_of_pages'].fillna(mean_no_of_pages)
bookshop['no_of_pages'].isna().sum()
```

```python
bookshop.info()
```

```python
median_height = bookshop['height'].median()
bookshop['height']=bookshop['height'].fillna(mean_no_of_pages)
bookshop['height'].isna().sum()
```

```python
median_height = bookshop['width'].median()
bookshop['width']=bookshop['width'].fillna(mean_no_of_pages)
bookshop['width'].isna().sum()
```

```python
bookshop['spine']=bookshop['spine'].fillna("UNKNOWN")
bookshop['spine'].isna().sum()
```

```python
median_weight = bookshop['weight'].median()
bookshop['weight']=bookshop['weight'].fillna(mean_no_of_pages)
bookshop['weight'].isna().sum()
```
