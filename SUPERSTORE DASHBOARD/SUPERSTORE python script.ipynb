```python
import pandas as pd
```

```python
superstore = pd.read_csv('Sample - Superstore.csv', encoding='ISO-8859-1')
superstore.head()
```

```python
row_id_index =superstore.set_index('Row ID')
row_id_index.head()
```

```python
superstore.columns
```

```python
superstore.info()
```

```python
superstore.isna().sum()
```

```python
superstore['Order Date'] = pd.to_datetime(superstore['Order Date'])
superstore['Ship Date'] = pd.to_datetime(superstore['Ship Date'])

superstore['Between OrdAndShip days'] = (superstore['Ship Date'] - superstore['Order Date']).dt.days

superstore[['Order Date', 'Ship Date', 'Between OrdAndShip days']].head()
```

```python
superstore.columns
```

```python
superstore.describe()
```

```python
superstore_year_month = superstore.groupby([
    superstore['Order Date'].dt.year,
    superstore['Order Date'].dt.month
]).agg({
    'Between OrdAndShip days': 'mean',
    'Quantity': 'sum',
    'Sales': 'sum',
    'Profit': 'sum'
})

superstore_year_month
```

```python
superstore_ship_mode_year_month = superstore.groupby([
    superstore['Ship Mode'],
    superstore['Order Date'].dt.year,
    superstore['Order Date'].dt.month
]).agg({
    'Between OrdAndShip days': 'mean',
    'Quantity': 'sum',
    'Sales': 'sum',
    'Profit': 'sum'
})

superstore_ship_mode_year_month
```

```python
superstore_geography_year_month = superstore.groupby([
    superstore['Country'],
    superstore['City'],
    superstore['State'],
    superstore['Order Date'].dt.year,
    superstore['Order Date'].dt.month
]).agg({
    'Between OrdAndShip days': 'mean',
    'Quantity': 'sum',
    'Sales': 'sum',
    'Profit': 'sum'
})

superstore_geography_year_month
```

```python
superstore_segment_year_month = superstore.groupby([
    superstore['Segment'],
    superstore['Order Date'].dt.year,
    superstore['Order Date'].dt.month
]).agg({
    'Between OrdAndShip days': 'mean',
    'Quantity': 'sum',
    'Sales': 'sum',
    'Profit': 'sum'
})

superstore_segment_year_month
```

```python
superstore_product_category_year_month = superstore.groupby([
    superstore['Category'],
    superstore['Sub-Category'],
    superstore['Order Date'].dt.year,
    superstore['Order Date'].dt.month
]).agg({
    'Between OrdAndShip days': 'mean',
    'Quantity': 'sum',
    'Sales': 'sum',
    'Profit': 'sum'
})

superstore_product_category_year_month
```

```python
superstore_product_name_year_month = superstore.groupby([
    superstore['Product Name'],
    superstore['Order Date'].dt.year,
    superstore['Order Date'].dt.month
]).agg({
    'Between OrdAndShip days': 'mean',
    'Quantity': 'sum',
    'Sales': 'sum',
    'Profit': 'sum'
})

superstore_product_name_year_month
```

```python
superstore_year_month.to_csv('superstore_year_month.csv',index=False)
```
